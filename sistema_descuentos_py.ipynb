```python
print("===SISTEMA DE DESCUENTO===")

edad = int(input("¿Cuántos años tienes?"))

estudiante = input("¿Eres estudiante?: (Si/No)").strip().capitalize()
deuda = input("¿Tienes alguna deuda? (Si/No)").strip().capitalize()

if estudiante not in ("Si", "No") or deuda not in ("Si","No"):
   print("Error debes escribir Si o No")

elif edad >=60:
     print("Resultado: Descuento de adulto mayor.")

elif estudiante =="Si" and deuda =="No":
     print("Resultado: Descuento estudiantil")

else:
     print("Resultado: No tienes descuento")
```
